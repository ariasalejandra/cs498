```python
## Homework 1
import pandas as pd
import numpy as np

df = pd.read_csv('YouTubeDataset_withChannelElapsed.csv')
```

```python
df['videoId'].duplicated().sum()
```

```python
df_unique = (
    df.groupby('videoId', as_index=False)
      .agg({
          'videoViewCount': 'max',   
          'videoLikeCount': 'max',
          'VideoCommentCount': 'max',
          'videoDislikeCount': 'max',
      })
)

df_unique['videoId'].duplicated().sum()
```

```python
df_unique
```

```python
# 1.Calculate the average number of likes per video.
avg_likes = df_unique['videoLikeCount'].mean()
avg_likes
```

```python
# 2. Compute the engagement rate for each video, where engagement rate is defined as the sum of likes, dislikes, 
# and comments divided by the total number of views for that video. 
# Finally, find the average engagement rate across all videos in the dataset.
df_unique['engagement_rate'] = (df_unique['videoLikeCount'] + df_unique['videoDislikeCount'] + df_unique['VideoCommentCount']) / df_unique['videoViewCount']
avg_engagement_rate = df_unique['engagement_rate'].mean()
avg_engagement_rate
```

```python
# 3. Each video contains 500 bytes of metadata and a 200 KB thumbnail. Calculate the combined storage size for a single video in megabytes (MB)
metadata_mb = 500 / (1024 * 1024)  
thumbnail_mb = 200 / 1024 
total_mb = metadata_mb + thumbnail_mb
total_mb
```

```python
# 4. Each time someone engages with a video, the video’s metadata (likes , dislikes, comments, and thumbnail) 
# is accessed and potentially transmitted. The total data transfer refers to the sum of all data accessed for all engagements. 
# Using the total number of engagements and combined storage size of a single video (You calculated in Question 3), 
# estimate the total data transfer required for all engagements with videos in the dataset. 

mb = 0.1958
engagements = df_unique['videoLikeCount'].sum() + df_unique['videoDislikeCount'].sum() + df_unique['VideoCommentCount'].sum()
total_data = engagements * mb
total_data
```

```python
# 5. Estimate the number of views per minute, assuming the dataset spans 5 years with views uniformly distributed over time.
total_views = df_unique['videoViewCount'].sum()
total_mins = 5 * 365 * 24 * 60 # this should be correct..... 5 years, 365 days per year, 24 hours per day, 60 minutes per hour..... yeah..... idk if we should consider leap years though
views_per_min = total_views / total_mins
views_per_min
```

```python
# 6. Estimate the amount of data accessed per minute to display video details to viewers.
data_per_min = views_per_min * mb
data_per_min
```
